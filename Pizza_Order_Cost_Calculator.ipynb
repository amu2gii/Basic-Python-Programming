````python
# User inputs for pizza size, number of toppings and delivery distance
pizza_size = input("Enter pizza size (small, large): ")
num_toppings = int(input("Enter number of toppings: "))
delivery_distance = float(input("Enter delivery distance (in kilometers): "))

# Use if and elif statements for determining cost of pizza
if pizza_size == "small":
    cost = 8
elif pizza_size == "large":
    cost = 12

else:
  print("Invalid pizza size. Defaulting the base cost to 0")
  base_cost = 0

# Calculating the cost of additional toppings($1 per toppings)
toppings_cost = num_toppings * 1

# Calculate the delivery fee
if delivery_distance <= 5:
    delivery_fee = 2
else:
    extra_miles = delivery_distance - 5
    delivery_fee = 2 + (extra_miles * 1)

# Calculate the total order cost
total_cost = cost + toppings_cost + delivery_fee

# Display the total order cost
print(f"Total order cost: ${total_cost:.2f}")
````
